```python
data_sorted = np.sort(df['Kms_Driven'])
cdf = np.arange(len(data_sorted)) / float(len(data_sorted))
plt.plot(data_sorted, cdf, marker='.', linestyle='none', color='orange')
plt.title("CDF of Transformed Selling_Price")
plt.show()
```

CORRELATION MATRIX BY USING HEAT MAP

```python
numeric_df = df.select_dtypes(include='number')
correlation_matrix = numeric_df.corr()
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix of Car Features')
plt.show()
```

- Year:
- A weak positive correlation with Price (0.24).
- A strong negative correlation with Driven (-0.52).
- Price:
- Shows a moderate correlation with Driven (-0.48).
- Slight negative correlation with other variables.
